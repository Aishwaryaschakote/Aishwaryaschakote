##student manager: store 5 students names in a list, converts them into a set to remove duplicates, stores each student with marks


```python
stud=['Rahul','shakili','aish','chinna', 'aish']
print("original_list:", stud)

stud_unique=set(stud)
print("unique_list:", stud_unique)

print("new dict")
stud_marks={}

print("taking marks for each student")
for stud in stud_unique:
  markss=int(input(f'enter marks for {stud}:'))
  stud_marks[stud]=markss

print("students scoring above 80: ")

for stud, marks in stud_marks.items():
  if marks>80:
    print(stud)
```
