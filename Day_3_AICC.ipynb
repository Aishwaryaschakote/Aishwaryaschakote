```python
fruits=['apple', 'banana', 'cherry']
fruits.append('orange')
print(fruits)

#insert at perticular position
fruits.insert(2,'grapes')
print(fruits)

#removing one item banana
fruits.remove('banana')
print(fruits)

#using pop to remove only last item
fruits.pop()
print(fruits)

#using update() to update a item in list
fruits[0]='pineapple'
print(fruits)
```
