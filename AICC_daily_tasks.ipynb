```python
#sum until user stops: keep asking the user to enter the nbrs and calculate the sum. stop only when the user enters 0

sum=0
while True:
  user_n=int(input("enter a nbr to be summed:"))
  if(user_n==0):
    break
  else:
    sum+=user_n
print(sum)
```

```python
#pattern
''' *
    * *
    * * *
    * * * *
'''
n=4
for i in range(1,n+1):
  for j in range(i):
    print("*", end=" ")
  print()
```
